```python
import matplotlib.pyplot as plt
import pickle
import numpy as np

def pickleRead(filename):
    with open(filename, "rb") as f:
        obj = pickle.load(f)
    return obj
```

```python
speed_vals = pickleRead("MeasureSpeedFromThrottleTest.pickle")
for _ in range(5): speed_vals[-1].append(1235)
throttle_vals = [0.05*i for i in range(21)]
speed_means = []
speed_stddiv = []
for i in speed_vals:
    speed_means.append(np.array(i)[len(i)//2:].mean())
    speed_stddiv.append(np.array(i)[len(i)//2:].std())
#speed_means[-1] = 1235
#speed_stddiv[-1] = 0.02
speed_means = np.array(speed_means)
speed_stddiv = np.array(speed_stddiv)
```

```python
print(throttle_vals)
print(speed_means)
print(speed_stddiv)
```

```python
Fig, ax = plt.subplots()
ax.plot(throttle_vals, speed_means)
print(np.take(np.array(speed_means), (0, 2, 10, 20)))
ax.plot(np.take(np.array(throttle_vals), (0, 2, 10, 20)), [0, 550, 1040, 1235])
#ax.plot(throttle_vals, -1235*np.exp2(-np.array(throttle_vals)*8)+1235)
#ax.plot(throttle_vals, -1235*np.exp(-np.array(throttle_vals)*4)+1235)
#ax.plot(throttle_vals, np.sqrt(np.array(throttle_vals)*2)*1235/np.sqrt(2))
plt.show()
```
